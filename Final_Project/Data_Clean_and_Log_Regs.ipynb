***Data Cleaning + Logistic Regression #2 - unsafe gun storage practices (loadulk2) ~ depression (ADDEPEV3)/poor mental health status (_MENT14D)***

```python
import numpy as np
import pandas as pd
import os 
import scipy.stats as stats
import statsmodels.formula.api as smf
```

```python
os.chdir(r"C:\Users\josep\Documents\GitHub\BIOS_584\Final_Project\data")

brfss_df = pd.read_csv("BRFSS_subset.csv")

brfss_df.columns.tolist()
```

**Clean Variables - firearm5, loadulk2, _ment14d, _racegr3, primins1, educa, income3, children** 

```python
#Sam is cleaning ADDEPEV3 and _AGE80

# Clean FIREARM5 - set Don't know, Refused, and BLANK to missing (NaN). 

df = brfss_df.copy()

df["firearm5"] = df["firearm5"].replace({7: np.nan, 9: np.nan})

print(df["firearm5"].value_counts(dropna=False))

# Clean loadulk2 - Don't know, Refused, and BLANK to missing (NaN). 
df["loadulk2"] = df["loadulk2"].replace({7: np.nan, 9: np.nan})

print(df["loadulk2"].value_counts(dropna=False))

# Clean _MENT14D - set Don't know, Refused, and BLANK to missing (NaN). 

df["_ment14d"] = df["_ment14d"].replace({9: np.nan})

print(df["_ment14d"].value_counts(dropna=False))

# Clean _RACEGR3 - set Don't know, Refused, and BLANK to missing (NaN). 
df["_racegr3"] = df["_racegr3"].replace({9: np.nan})

print(df["_racegr3"].value_counts(dropna=False))

# Clean PRIMINS1 - set Don't know, Refused, and BLANK to missing (NaN). 
df["primins1"] = df["primins1"].replace({77:np.nan, 99: np.nan})

print(df["primins1"].value_counts(dropna=False))

# Clean EDUCA - set Refused and BLANK to missing (Nan). 
df["educa"] = df["educa"].replace({9:np.nan})

print(df["educa"].value_counts(dropna=False))

# Clean Income - set Don't know, Refused and BLANK to missing (Nan). 
df["income3"] = df["income3"].replace({77:np.nan, 99: np.nan})

print(df["income3"].value_counts(dropna=False))

# Clean children - set Refused and BLANK to missing (Nan). 
df["children"] = df["children"].replace({99: np.nan})

print(df["children"].value_counts(dropna=False))
```

***Logistic Regression (2a) Unsafe Gun Storage Practices ~ Depression + covariates***

```python
#Outcome: loadulk2 (unsafe storage)
#main predictor: addepev3 (0/1 depression)
#covariates: _age80, sexvar, _racegr3, educa, income3, primins1, children

# keep only the respondents who have guns 
gun_df = df[df["firearm5"] == 1].copy()
print(gun_df["firearm5"].value_counts(dropna=False))

vars_2a = ["loadulk2", "addepev3", "_age80", "sexvar", "_racegr3", "educa", "income3", "primins1", "children"] #all vars needed for model 2a

m2a = gun_df[vars_2a].dropna().copy() # drop rows with any missings in these variables

#Logistirc Regression
logreg_2a = "loadulk2 ~ addepev3 + C(_age80) + C(sexvar) + C(_racegr3) + C(educa) + C(income3) + C(primins1) + C(children)"

model_2a = smf.logit(formula=logreg_2a, data=m2a) #https://www.statsmodels.org/stable/generated/statsmodels.formula.api.logit.html
result_2a = model_2a.fit()

print(result_2a.summary())
```

***Logistic Regression (2b) Unsafe Gun Storage Practices ~ Days of Poor Mental Health + covariates***

```python
#Outcome: loadulk2 (unsafe storage)
#main predictor: menthlth (continuous; # days of poor mental health)
#covariates: _age80, sexvar, _racegr3, educa, income3, primins1, children

# keep only the respondents who have guns 
gun_df = df[df["firearm5"] == 1].copy()
print(gun_df["firearm5"].value_counts(dropna=False))

vars_2b = ["loadulk2", "menthlth", "_age80", "sexvar", "_racegr3", "educa", "income3", "primins1", "children"] #all vars needed for model 2a

m2b = gun_df[vars_2b].dropna().copy() # drop rows with any missings in these variables

#Logistirc Regression
logreg_2b = "loadulk2 ~ menthlth + C(_age80) + C(sexvar) + C(_racegr3) + C(educa) + C(income3) + C(primins1) + C(children)"

model_2b = smf.logit(formula=logreg_2b, data=m2b) #https://www.statsmodels.org/stable/generated/statsmodels.formula.api.logit.html
result_2b = model_2b.fit()

print(result_2b.summary())
```
