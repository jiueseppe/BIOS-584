## Import Data

<font size='4'>

* Three lists named `study_id_ls`, `measure_date_ls`, and `duration_ls` have been initialized.
* Write a for loop to extract and append the values of `study_id`, `measure_date`, and `duration` to `study_id_ls`, `measure_date_ls`, and `duration_ls`, respectively.

```python
study_id_ls = []
measure_date_ls = []
duration_ls = []

folder = "data/Quiz4"

file_names = ["data_1.txt", "data_2.txt", "data_3.txt", "data_4.txt", "data_5.txt", "data_6.txt"]

for file_name in file_names: 
    path = os.path.join(folder, file_name)
    temp_df = pd.read_csv(path)

    # extracts only the value from first row
    study_id_value = temp_df["study_id"][0]
    measure_date_value = temp_df["measure_date"][0]
    duration_value = temp_df["duration"][0]
    
    # adds the value to its specfic list
    study_id_ls.append(study_id_value)
    measure_date_ls.append(measure_date_value)
    duration_ls.append(duration_value)

#check
print(study_id_ls)
print(measure_date_ls)
print(duration_ls)

```

<font size='4'>

* Create a new Pandas Dataframe combining three lists and name it `input_df`.

```python
input_df = pd.DataFrame({"study_id": study_id_ls, "measure_date": measure_date_ls, "duration": duration_ls})

input_df.shape
```

<font size='4'>

* Show or print `input_df`

```python
input_df
```

## Data Manipulation

<font size='4'>

* Reorder the dataframe by `measure_date` (earliest to latest) and then by `duration` (shortest to longest).

```python
input_df = input_df.sort_values(by=["measure_date", "duration"], ascending=[True, True])

input_df
```

<font size='4'>

* Create a new column called `relative_time` by calculating the time difference in days between each of the measure date and the earliest one.

```python
input_df["measure_date"] = pd.to_datetime(input_df["measure_date"]) #to convert to real actual dates

earliest_date = input_df["measure_date"].min() 
print(earliest_date) #should be 1/1/25

input_df["relative_time"] = (input_df["measure_date"] - earliest_date)
print(input_df["relative_time"])
```

<font size='4'>

* Reset the index column and remove the old index.

```python
input_df = input_df.reset_index(drop=True)
```
